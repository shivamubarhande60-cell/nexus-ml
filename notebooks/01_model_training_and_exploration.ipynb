## 1. Dataset Generation and Summary Statistics

## 2. Train Random Forest Classifier and Evaluate

```python
preprocessor = ScreeningPreprocessor()
X_scaled = preprocessor.fit_transform(df)
y = df['screening_label'].values

rf = StructuredScreeningModel(model_type='rf')
metrics = rf.train_and_evaluate(X_scaled, y, feature_names=FEATURE_COLUMNS)
for k, v in metrics.items():
    print(f"{k.capitalize():<15}: {v:.4f}")
```

## 3. Top Contributing Features (Feature Importances)

```python
importances = rf.get_feature_importances()
feat_df = pd.DataFrame(list(importances.items()), columns=['Feature', 'Importance'])
feat_df.head(10)
```
